# Stemming & its types


Stemming is the process in which we reduce the word to get stem of the word. It is a simple and fast method for text preprocessing, but it may not always produce the correct root of a word. There are different types of stemming algorithms, including:

1. Porter Stemmer: This is the most commonly used stemming algorithm. It uses a set of rules to remove suffixes from words. For example, it will reduce the word "running" to "run" and "happiness" to "happi".

2. regexpStemmer: This is a more flexible approach that uses regular expressions to define the stemming rules. It allows for more complex patterns and can be customized to fit specific needs.

3. Snowball Stemmer: This is an improvement over the Porter stemmer. It is more aggressive and can handle a wider range of words. It is also more accurate in terms of producing the correct root of a word, still it is not perfect and may not always produce the correct root of a word. For example, the stemmer may reduce the word "history" to "histori", which is not a valid word.

```python
## Classification Problem
## Comments of product is a positive review or negative review
## Reviews ---- > eating, eat, eaten [going, gone, goes] --- > go

words = [
    "history","eating", "eat", "eaten",
    "going", "gone", "goes",
    "playing", "played", "plays",
    "studies", "studying", "studied",
    "flies", "flying", "flown",
    "happily", "happiness",
    "liked", "liking",
    "amazing", "amazingly",
    "useful", "useless"
]       
```

## Porterstemmer

```python
from nltk.stem import PorterStemmer
```

```python
stemming = PorterStemmer()
```

```python
for word in words:
    print(f"{word} ---> {stemming.stem(word)}")
```

```python
stemming.stem('congratulations')
```

It is the major disadvantage of stemming that it may not always produce the correct root of a word. For example, the stemmer may reduce the word "congratulations" to "congratul", which is not a valid word.

### RegexpStemmer class

NLTK has RegexpStemmer class with the help of which we can easily implement Regular
Expression Stemmer algorithms. It basically takes a single regular expression and removes any
prefix or suffix that matches the expression.

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer=RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

```python
reg_stemmer.stem('comfortable')
```
