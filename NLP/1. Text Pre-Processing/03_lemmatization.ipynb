# Wordnet Lemmatizer

Lemmatization technique isplike stemming. The output we will get after lemmatization is called
lemma', which is a root word rather than root stem, the output of stemming. After lemmatization,
we will be getting a valid word that means the same thing.

NLTK provides WordNetLemmatizer class which is a thin wrapper around the wordnet corpus. This
class uses morphy() function to the WordNet CorpusReader class to find a lemma. Let us
understand it with an example -

it is better to use lemmatization when we want to get a valid word as output. For example, the word "better" is the comparative form of "good". When we lemmatize "better", we get "good" as the lemma.

But it is slower than stemming because it uses a dictionary lookup to find the lemma. So, if you want to get a valid word as output, use lemmatization. If you want to get a root stem, use stemming.

```python
## Q&A, chatbots,text summarization
from nltk.stem import WordNetLemmatizer
```

```python
import nltk

nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
lemmatizer=WordNetLemmatizer()
```

```python
'''
POS- Noun-n
verb-v
adjective-a
adverb-r
'''
lemmatizer.lemmatize("going")
```

```python
lemmatizer.lemmatize("going" , pos='v')
```

```python
words = ['running', 'ran', 'easily', 'fairly' , 'goes', 'sportingly', 'history' , 'better', 'best', 'worse', 'worst', 'good', 'bad', 'happy', 'happily', 'sad', 'sadly']
```

```python
for word in words:
    print(f"{word} ---> {lemmatizer.lemmatize(word)}")
```

```python
for word in words:
    print(f"{word} ---> {lemmatizer.lemmatize(word , pos='v')}")
```
