## 1. Introduction & Setup

In NLP, **Tokenization** is the process of breaking down text (a corpus or paragraph) into smaller semantic units called **tokens** (such as sentences or words).

### Core NLP Terminologies:
* **Corpus:** The complete collection of raw text documents or paragraphs.
* **Documents:** Individual sentences or logical text units within a corpus.
* **Vocabulary:** The set of unique words/tokens present across the corpus.
* **Words / Tokens:** Individual units of meaning used for downstream tasks (vectorization, stemming, lemmatization, etc.).

## 3. Sentence Tokenization (`sent_tokenize`)

### What does it do?
`sent_tokenize` breaks a paragraph/corpus into a list of individual sentences (often called **documents** in NLP).

### How does it work?
Under the hood, it uses NLTK's **`PunktSentenceTokenizer`**, an unsupervised machine-learning tokenizer trained to detect sentence boundaries, abbreviations, and punctuation.

* It splits on sentence boundaries: full stop (`.`), exclamation (`!`), question mark (`?`).
* It handles edge cases like abbreviations (e.g., "Dr.", "Mr.") and decimal numbers (e.g., "3.14") to avoid incorrect splits.

```python
from nltk.tokenize import sent_tokenize

# Convert paragraph into sentences (documents)
documents = sent_tokenize(corpus)
print("Type of documents:", type(documents))
print("Total sentences found:", len(documents))
print("\nList of documents:")
documents
```

```python
# Iterating through documents (sentences)
for index, sentence in enumerate(documents, start=1):
    print(f"Sentence {index}: {sentence}")
```

## 4. Word Tokenization (`word_tokenize`)

### Why convert to words?
In Natural Language Processing:
1. Every word carries individual semantic meaning and importance.
2. Downstream operations—such as **Stopword removal**, **Stemming**, **Lemmatization**, **Bag of Words (BoW)**, **TF-IDF**, and **Word2Vec**—operate primarily at the word/token level.

### Key Observation with `word_tokenize`:
* Punctuation marks (like `,`, `.`, `!`) are separated into their own tokens.
* This is crucial for tasks like sentiment analysis, where punctuation can change the meaning of a sentence.


```python
from nltk.tokenize import word_tokenize

# Tokenize the entire corpus into words
words = word_tokenize(corpus)
print(f"Total tokens: {len(words)}")
print(words)
```

```python
# Word tokenization per sentence (document)
for i, sentence in enumerate(documents, start=1):
    tokenized_sentence = word_tokenize(sentence)
    print(f"Sentence {i} words ({len(tokenized_sentence)} tokens):")
    print(tokenized_sentence)
    print()
```

## 5. WordPunctTokenizer

### How it behaves:
`WordPunctTokenizer` splits text into alphabetic and non-alphabetic tokens using regular expression pattern matching (`\w+|[^\w\s]+`).

### Crucial Difference vs `word_tokenize`:
* Look at **`Krish Naik's`**:
* In `word_tokenize`: `['people', "'s"]`
    * In `WordPunctTokenizer`: `['people', "'", 's']` (the single quote `'` and the letter `s` are split into separate tokens).
* Every punctuation character is strictly isolated as its own standalone token.

```python
from nltk.tokenize import WordPunctTokenizer

# Initialize WordPunctTokenizer
wpt = WordPunctTokenizer()

# Tokenize corpus
wpt_tokens = wpt.tokenize(corpus)
print(f"Total tokens: {len(wpt_tokens)}")
print(wpt_tokens)
```

## 6. TreebankWordTokenizer

### Background:
This tokenizer follows the conventions established by the **Penn Treebank corpus** (standard in computational linguistics and syntax treebanks).

### Key Differences & Quirks (Highlight from Lecture):
1. **Handling of Periods (`.`):**
   * If a period is followed immediately by a newline without trailing space (like `tutorials.\nPlease`), `TreebankWordTokenizer` keeps the period attached to the token: `'tutorials.'`.
   * However, at the absolute end of the sentence/corpus (`expert in NLP.`), the period is split into a separate token `'.'`.
2. **Contractions:**
   * It separates standard contractions according to Penn Treebank rules.

```python
from nltk.tokenize import TreebankWordTokenizer

# Initialize TreebankWordTokenizer
tb_tokenizer = TreebankWordTokenizer()

# Tokenize corpus
tb_tokens = tb_tokenizer.tokenize(corpus)
print(f"Total tokens: {len(tb_tokens)}")
print(tb_tokens)
```

## 7. Side-by-Side Tokenizer Comparison Matrix

Let's test all three word tokenizers on the exact phrase: `"Krish Naik's NLP tutorials."`


```python
sample_text = "Krish Naik's NLP tutorials."

print("Original Text:        ", sample_text)
print("-" * 65)
print("1. word_tokenize:     ", word_tokenize(sample_text))
print("2. WordPunctTokenizer: ", WordPunctTokenizer().tokenize(sample_text))
print("3. TreebankTokenizer:  ", TreebankWordTokenizer().tokenize(sample_text))
```

### Summary Comparison Table:

| Feature / Tokenizer | `sent_tokenize` | `word_tokenize` | `WordPunctTokenizer` | `TreebankWordTokenizer` |
| :--- | :--- | :--- | :--- | :--- |
| **Output Type** | Sentences (`list[str]`) | Words & Punctuation | Words & Punctuation | Penn Treebank tokens |
| **Apostrophe (`'s`)** | Preserved in sentence | Keeps `'s` together | Splits into `'` and `s` | Keeps `'s` together |
| **Punctuation** | Boundary delimiter | Separate tokens | Separate tokens | Depends on position/newline |
| **Primary Use Case** | Document segmentation | General NLP pipeline | Heavy regex/punct tasks | Linguistic syntax parsing |


### Key Differences:

| Criteria | **NLTK (Natural Language Toolkit)** | **spaCy** |
| :--- | :--- | :--- |
| **Philosophy & Target** | Academic, research-oriented, teaching | Industry-grade, production-ready applications |
| **Implementation** | Pure Python algorithms | Written in high-performance **Cython** (C-speed) |
| **Algorithms** | Multiple choices for each task (e.g., 5+ tokenizers) | One state-of-the-art opinionated algorithm per task |
| **Data Structure** | Operates on plain strings and lists of strings | Object-oriented (`Doc`, `Token`, `Span` objects) |
| **Pipeline Integration** | Manual step-by-step function calling | Unified, customizable processing pipeline |
| **Deep Learning & Vectors**| Basic rule-based / statistical models | Built-in pre-trained Transformer / Word vector models |
| **Execution Speed** | Moderate / Slower on large datasets | Extremely fast (benchmark leader) |
