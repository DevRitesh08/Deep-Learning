```python
messages = pd.read_csv('SMSSpam.txt', sep='\t', names=["label", "message"])
```

## Data Cleaning & Preprocessing

```python
import re
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

**apply lemmatization soon**

```python
corpus = []

for i in range(0, len(messages)):
    review = re.sub('[^a-zA-Z]', ' ', messages['message'][i])       # Remove all the special characters and numbers
    review = review.lower()
    review = review.split()
    
    review = [lemmatizer.lemmatize(word) for word in review if not word in set(stopwords.words('english'))]
    review = ' '.join(review)
    corpus.append(review)
```

```python
corpus
```

## Create the Bag of Words model

```python
from sklearn.feature_extraction.text import CountVectorizer

# n_grams_range = (1, 2)  # This will create both unigrams and bigrams , (1,1) for unigrams only, (2,2) for bigrams only, (1,3) for unigrams, bigrams and trigrams, etc.
cv = CountVectorizer(max_features=2500 , binary=True)     # it has parameters like  binary, lowercase, stop_words, max_features, ngram_range, etc. which can be used to customize the vectorization process. (no need to be done manually as we did above)
```

for binary BOW model, we can set the parameter binary=True in CountVectorizer. This will create a binary representation of the text data, where each feature is either 0 or 1, indicating the absence or presence of a word in the document.

```python
cv.vocabulary_
```

```python
X = cv.fit_transform(corpus).toarray()
```

```python
X.shape
```

## N - Grams

n_grams are a contiguous sequence of n items from a given sample of text or speech. In the context of text data, n-grams are typically sequences of words or characters. For example, in the sentence "I love natural language processing", the 2-grams (bigrams) would be: "I love", "love natural", "natural language", "language processing". N-grams can be useful for capturing context and relationships between words in text data.

```python
cv = CountVectorizer(
    max_features=500,
    binary=True,
    ngram_range=(3, 3)
)

X=cv.fit_transform(corpus).toarray()
```

```python
cv.vocabulary_
```

for accuracy go from 1 to 3 n-grams, but for speed and memory usage, go for 1 to 2 n-grams.
